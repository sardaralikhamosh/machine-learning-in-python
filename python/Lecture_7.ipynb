## Variable Scope

Variable scope means where a variable can be used in a program.  
There are mainly two types of scope:

1. Local Variable

A local variable is created inside a function and can only be used inside that function.

2. Global Variable

A global variable is created outside the function and can be used anywhere in the program.


```python
def my_function():
    x = 10      # local variable
    print("Value inside function:", x)

my_function()
```

```python
x = 20   # global variable

def my_function():
    print("Value inside function:", x)

my_function()
print("Value outside function:", x)
```

### Question: Mini ATM System Using Functions

Create a Mini ATM System using functions. The ATM should start with a balance of Rs. 10,000. Create separate functions for Deposit, Withdraw, and Check Balance. Use a while loop to display the menu until the user selects Exit. Update the balance after each transaction and display an "Insufficient Balance" message if the withdrawal amount is greater than the available balance. Print "Thank you for using our ATM!" when the user exits the program.


```python
def deposit(balance, amount):
    balance += amount
    print("Deposit Successful!")
    return balance


def withdraw(balance, amount):
    if amount <= balance:
        balance -= amount
        print("Withdrawal Successful!")
    else:
        print("Insufficient Balance!")
    return balance


def check_balance(balance):
    print("Current Balance:", balance)


balance = 10000

while True:
    print("\n===== ATM MENU =====")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        amount = int(input("Enter deposit amount: "))
        balance = deposit(balance, amount)

    elif choice == "2":
        amount = int(input("Enter withdrawal amount: "))
        balance = withdraw(balance, amount)

    elif choice == "3":
        check_balance(balance)

    elif choice == "4":
        print("Thank you for using our ATM!")
        break

    else:
        print("Invalid Choice!")
```

## Map Function

The map() function applies a function to every element of a list (or iterable).

- map(function, iterable)


```python
string = ["apple","orange","manogo","graps"]
length = map(len, string)
print(list(length))
```

```python
num1 = [1,2,3,4,5,6]
double = list(map(lambda n : n*n,num1))
print(double)
```

```python
celsius = [20, 25, 30]

fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))

print(fahrenheit)
```

### Different Modes

- "r+": Open for reading and writing. The stream is positioned at the beginning of the file.
- "w+": Open for reading and writing. The file is created if it does not
  exist, otherwise it is truncated. The stream is positioned at
  the beginning of the file.
- "a+": Open for reading and writing. The file is created if it does not
  exist. The stream is positioned at the end of the file. Subse-
  quent writes to the file will always end up at the then current
  end of file, irrespective of any intervening fseek(3) or similar.

## Exception Handling

Exception handling means handling runtime errors so the program does not crash.

Instead of stopping the program, we control the error.

Python provides try and except for this purpose.


```python
# try:
    #code that may cause error
# except:
    #code that runs if error occurs
```

```python
a = 4
b = "0"
try:
    print(a/b)
except Exception as e:
    print(e)
```

```python
a = 5
b = 0
try:
    print(a/b)
    num = int(input("enter a number"))
    print(num)
# except Exception as e:
#     print("You can not divide a number with zero:",e)
except ZeroDivisionError:
    print("You can not divide a number with zero:")
except ValueError as e:
    print("error:",e)
print("Bye")
```
