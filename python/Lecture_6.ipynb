#### 1. Online Quiz

Create a quiz with 5 questions. Ask one question at a time, check the answer using if-else, and keep track of the score. Display the final score at the end.


````python
score = 0
question = 1

while question <= 5:

    if question == 1:
        answer = input("Q1. What is the capital of Pakistan? ")
        if answer.lower() == "islamabad":
            print("Correct!")
            score += 1
        else:
            print("Wrong!")

    elif question == 2:
        answer = input("Q2. How many days are there in a week? ")
        if answer == "7":
            print("Correct!")
            score += 1
        else:
            print("Wrong!")

    elif question == 3:
        answer = input("Q3. Which language is used for AI and Machine Learning? ")
        if answer.lower() == "python":
            print("Correct!")
            score += 1
        else:
            print("Wrong!")

    elif question == 4:
        answer = input("Q4. What is 10 + 5? ")
        if answer == "15":
            print("Correct!")
            score += 1
        else:
            print("Wrong!")

    elif question == 5:
        answer = input("Q5. Which keyword is used for a loop that runs while a condition is True? ")
        if answer.lower() == "while":
            print("Correct!")
            score += 1
        else:
            print("Wrong!")

    question += 1

print("\nQuiz Finished!")
print("Your Score:", score, "/5")
````

### Task 3

#### Restaurant Billing System

Create a **Restaurant Billing System**.

#### Menu

| Item   | Price (Rs.) |
| ------ | ----------: |
| Pizza  |        1200 |
| Burger |         500 |
| Fries  |         300 |
| Juice  |         250 |

Keep asking the user to enter an item name until they type **"finish"**.

### Requirements

- Store all ordered items in a list.
- Calculate the total bill.
- Apply **5% tax** to the total amount.
- Print a restaurant-style receipt showing:
  - All ordered items with their prices.
  - Subtotal
  - Tax (5%)
  - Final Total
- If the user enters an item that is not in the menu, display **"Item not available"**.

#### Sample Output

```text
----------- MENU -----------

Pizza   : Rs.1200
Burger  : Rs.500
Fries   : Rs.300
Juice   : Rs.250

Enter item (or finish): pizza
Enter item (or finish): fries

Enter item (or finish): finish

--------- RECEIPT ---------

--------------------------
```

````python
items = []
prices = []
print("\nMenu")
print("Pizza  = Rs.1200")
print("Burger = Rs.500")
print("Fries  = Rs.300")
print("Juice  = Rs.250")

while True:
    order = input("\nEnter Item (or finish): ").lower()

    if order == "finish":
        break

    elif order == "pizza":
        items.append("Pizza")
        prices.append(1200)
        print(f"{order} is added")

    elif order == "burger":
        items.append("Burger")
        prices.append(500)
        print(f"{order} is added")

    elif order == "fries":
        items.append("Fries")
        prices.append(300)
        print(f"{order} is added")

    elif order == "juice":
        items.append("Juice")
        prices.append(250)
        print(f"{order} is added")

    else:
        print(f"{order} not available.")

print("\n========== RECEIPT ==========")

i = 0
subtotal = 0

while i < len(items):
    print(items[i], "........ Rs.", prices[i])
    subtotal += prices[i]
    i += 1

tax = subtotal * 0.05
total = subtotal + tax

print("------------------------------")
print("Subtotal :", subtotal)
print("Tax (5%) :", tax)
print("Total    :", total)

print("\nThank You!")
print("Visit Again :)")
````

### Mini ATM System

Create a Mini ATM System in Python. The ATM should start with a balance of Rs. 10,000. Display a menu with the following options: Deposit, Withdraw, Check Balance, and Exit. Use a while loop to keep showing the menu until the user selects Exit. Update the balance after each deposit or withdrawal, check for sufficient balance before withdrawing, and display the current balance when requested. Show an "Invalid Choice" message for incorrect input. As a bonus, use a for loop to print "Thank you for using our ATM" three times after the user exits the program.


````python
balance = 10000

while True:
    print("\n===== ATM MENU =====")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Deposit Successful!")
        print("Current Balance:", balance)

    elif choice == "2":
        amount = float(input("Enter withdrawal amount: "))

        if amount <= balance:
            balance -= amount
            print("Withdrawal Successful!")
            print("Current Balance:", balance)
        else:
            print("Insufficient Balance!")

    elif choice == "3":
        print("Current Balance:", balance)

    elif choice == "4":
        print("\nExiting ATM...")
        break

    else:
        print("Invalid Choice!")

print("\nThank you for using our ATM!")
````
