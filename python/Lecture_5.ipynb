## Dictionary in Python

Dictionaries are used to store data values in key : value pairs  
They are unordered, mutable(changeable) & don’t allow duplicate keys  
dict[”key”] = “value” #to assign or add new


```python
info = {"name":"ali"}
print(info)
print(type(info))
```

```python
info["name"] = "ahmad"
info["age"] = 23
print(info)
```

```python
student ={"name": "ali", 
        "age" :23,
        "cgpa": 3.5,
        "subject": {
            "phy" :67,
            "cham" :78,
            "math" :23
        }
    }
print(student)
student["subject"]
```

## Nested Dictionaries


```python
info = {
    "name" :"ali",
    "age":23,
    "subject": {
        "phy":67,
        "chem":89,
        "maths":90
    },
    ("topics",): "dict",
    "is_adult" : True
}
print(info["subject"]["chem"])
```

#### Question 5 – Online Shopping System

Create a dictionary where each product has a nested dictionary containing Price, Quantity, and Category.

Ask the user to enter a product name.

If the product exists and the quantity is greater than 0, print "Product Added to Cart".
Otherwise, print "Out of Stock".
If the product does not exist, print "Product Not Found".


```python
products = {
    "Laptop": {
        "Price": 120000,
        "Quantity": 5,
        "Category": "Electronics"
    },

    "Mouse": {
        "Price": 1500,
        "Quantity": 10,
        "Category": "Accessories"
    },

    "Keyboard": {
        "Price": 3000,
        "Quantity": 0,
        "Category": "Accessories"
    }
}

product = input("Enter Product Name: ")

if product in products:

    if products[product]["Quantity"] > 0:
        print("Product Added to Cart")
        print("Price:", products[product]["Price"])
        print("Category:", products[product]["Category"])

    else:
        print("Out of Stock")

else:
    print("Product Not Found")
```

#### Question 6 – Bank Account System

Create a dictionary containing three bank accounts. Each account should have a nested dictionary containing Account Holder, Balance, and Account Type.

Ask the user to enter an account number.

If the account exists and the balance is greater than 1000, print "Transaction Allowed".
Otherwise, print "Insufficient Balance".
If the account does not exist, print "Invalid Account Number".


```python
accounts = {
    "1001": {
        "Account Holder": "Ali",
        "Balance": 5000,
        "Account Type": "Saving"
    },

    "1002": {
        "Account Holder": "Ahmad",
        "Balance": 800,
        "Account Type": "Current"
    },

    "1003": {
        "Account Holder": "Sara",
        "Balance": 12000,
        "Account Type": "Saving"
    }
}

account_number = input("Enter Account Number: ")

if account_number in accounts:

    if accounts[account_number]["Balance"] > 1000:
        print("Transaction Allowed")

        print("Account Holder:", accounts[account_number]["Account Holder"])
        print("Balance:", accounts[account_number]["Balance"])
        print("Account Type:", accounts[account_number]["Account Type"])

    else:
        print("Insufficient Balance")

else:
    print("Invalid Account Number")
```

## Set Methods

- set.add( el ) #adds an element
- set.remove( el ) #removes the elem an
- set.clear( ) #empties the set
- set.pop( ) #removes a random value
- set.union( set2 ) #combines both set values & returns new
- set.intersection( set2 ) #combines common values & returns new


```python
set1 = {1,4,2 ,3}
set2 = {1,5,6}
print(set1.union(set2))
print(set1.intersection(set2))
```
