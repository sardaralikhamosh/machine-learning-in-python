## Rules for Identifiers

- Identifiers can only contain letters (A-Z, a-z), digits (0-9), and underscores (\_).
- No special symbols like @, $, or % are allowed.
- An identifier can have numbers in it (e.g., user1), but it cannot begin with a number (e.g., 1user is invalid).
- You cannot use Python's built-in keywords (like if, else, while, def, class, etc.) as identifier names.
- Identifier can be of any length.

## Types of Operators

An operator is a symbol that performs a certain operation between operands.

- Arithmetic Operators ( + , - ,\*, / , % ,\*\* )
- Relational / Comparison Operators ( == , != , > , < , >= , <= )
- Assignment Operators ( = , +=, -= ,\*= , /= , %= ,\*\*= )
- Logical Operators OR, AND, NOT


```python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a >= b)
print(a < b)
print(a <= b)
```

```python
x = 10
x += 5
print(x)
x -= 2
print(x)
x *= 2
print(x)
x /= 2
print(x)
x **= 2
print(x)
```

## Type Conversion

```python
tupp = ("a","c","a","b","d")
listt= list(tupp)
listt.sort()
print(listt)
print(type(listt))
```
