## List Slicing

Similar to String Slicing  
list_name[ starting_idx : ending_idx ] #ending idx is not included


```python
marks = [87, 64, 33, 95, 76,2,3,4,5,6,7,89,23,34]
print(marks[1:4])
print(marks[:4])
print(marks[1:])
```

## List Methods

- list.append(el) #adds one element at the end
- list.insert( idx, el ) #insert element at index
- list.sort( ) #sorts in ascending order
- list.reverse( ) #reverses list
- list.sort( reverse=True ) #sorts in descending order
- list.remove(el) #removes first occurrence of element
- list.pop( idx ) #removes element at idx


```python
list = [5,4,2,3,2,6,2,7]
list.remove(2)
list.pop(4)
# list.insert(2, 90)
# list.sort(reverse= True)
# list.reverse()
# list.remove(2)
# list.pop(5)
list.reverse()
print(list)
```

## Tuples in Python

A built-in data type that lets us create immutable sequences of values.


```python
t1 = (1,2,3,4,5)
print(type(t1))
print(t1)
```

```python
tup1 = (1,2, 3 ,2,2,4 ,6)
print(tup1.index(3))
print(tup1.count(2))
```

## Tuple Methods

- tup.index( el ) #returns index of first occurrence
- tup.count( el ) #counts total occurrences
